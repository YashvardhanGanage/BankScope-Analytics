```python

from pathlib import Path
import pandas as pd

DATA_PATH = Path("../data/raw/indian_banking_transactions.csv")

# Expected schema based on the sample
expected_columns = [
    "transaction_id", "customer_id", "transaction_date",
    "transaction_time", "account_type", "transaction_type",
    "transaction_amount", "transaction_direction",
    "account_balance", "merchant_category", "state",
    "credit_score", "has_loan", "loan_type", "emi_amount",
    "transaction_status", "channel", "kyc_status",
    "is_fraud", "transaction_hour"
]

# Read the complete dataset
df = pd.read_csv(DATA_PATH, low_memory=False)

print("DATASET OVERVIEW")
print("-" * 45)
print("Rows:", f"{len(df):,}")
print("Columns:", len(df.columns))
print("Duplicate transaction IDs:", df["transaction_id"].duplicated().sum())
print("Duplicate complete rows:", df.duplicated().sum())

print("\nSCHEMA CHECK")
print("Missing expected columns:",
      sorted(set(expected_columns) - set(df.columns)))
print("Unexpected columns:",
      sorted(set(df.columns) - set(expected_columns)))

print("\nMISSING VALUES")
missing = df.isna().sum()
missing_report = pd.DataFrame({
    "missing_count": missing,
    "missing_pct": (missing / len(df) * 100).round(2)
}).sort_values("missing_pct", ascending=False)
display(missing_report)

print("\nDATA TYPES")
display(df.dtypes.astype(str).rename("dtype").to_frame())

print("\nNUMERIC SUMMARY")
display(df[[
    "transaction_amount", "account_balance",
    "credit_score", "emi_amount", "transaction_hour"
]].describe().T)

print("\nDATE RANGE")
dates = pd.to_datetime(df["transaction_date"], errors="coerce")
print("Earliest date:", dates.min())
print("Latest date:", dates.max())
print("Invalid dates:", dates.isna().sum())
```

```python

categorical_columns = [
    "account_type",
    "transaction_type",
    "transaction_direction",
    "merchant_category",
    "transaction_status",
    "channel",
    "kyc_status",
    "has_loan",
    "is_fraud"
]

for column in categorical_columns:
    print(f"\n{column}")
    display(df[column].value_counts(dropna=False).head(20).to_frame("count"))

print("\nNUMERIC VALIDATION")
print("Amounts <= 0:", (df["transaction_amount"] <= 0).sum())
print("Negative balances:", (df["account_balance"] < 0).sum())
print("Credit scores outside 300–900:",
      (~df["credit_score"].between(300, 900)).sum())
print("Hours outside 0–23:",
      (~df["transaction_hour"].between(0, 23)).sum())

print("\nLOAN CONSISTENCY")
display(pd.crosstab(
    df["has_loan"],
    df["loan_type"].isna(),
    rownames=["has_loan"],
    colnames=["loan_type is missing"]
))
```

```python

# BankScope Analytics — targeted quality checks

# 1. Investigate loan-type consistency
loan_issues = df[
    (df["has_loan"] == 1) & (df["loan_type"].isna())
]

print("Customers/transactions marked as having a loan")
print("but missing loan_type:", len(loan_issues))

print("\nLoan types for records with has_loan = 1:")
display(
    df.loc[df["has_loan"] == 1, "loan_type"]
      .value_counts(dropna=False)
)

# 2. Check date distribution by year
df["transaction_date"] = pd.to_datetime(
    df["transaction_date"], errors="coerce"
)

print("\nTransactions by year:")
display(
    df["transaction_date"]
      .dt.year.value_counts()
      .sort_index()
      .rename_axis("year")
      .to_frame("transaction_count")
)

print("\nLatest 10 transaction dates:")
display(
    df["transaction_date"]
      .value_counts()
      .sort_index()
      .tail(10)
      .rename_axis("date")
      .to_frame("transaction_count")
)

# 3. Inspect extreme transaction amounts
print("\nTransaction amount percentiles:")
display(
    df["transaction_amount"]
      .quantile([0.50, 0.90, 0.95, 0.99, 0.999])
      .rename("amount")
      .to_frame()
)

print("\nTransactions above ₹1 lakh:", 
      (df["transaction_amount"] > 100_000).sum())

print("Transactions above ₹10 lakh:",
      (df["transaction_amount"] > 1_000_000).sum())

# 4. Verify binary fields
print("\nhas_loan unique values:", sorted(df["has_loan"].unique()))
print("is_fraud unique values:", sorted(df["is_fraud"].unique()))

# 5. Compare transaction status and direction
print("\nStatus by transaction direction:")
display(pd.crosstab(
    df["transaction_direction"],
    df["transaction_status"],
    margins=True
))
```

```python

from pathlib import Path
import pandas as pd

RAW_PATH = Path("../data/raw/indian_banking_transactions.csv")
PROCESSED_DIR = Path("../data/processed")
REPORT_DIR = Path("../reports")

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
REPORT_DIR.mkdir(parents=True, exist_ok=True)

# Reload the original data so the raw source remains unchanged
clean_df = pd.read_csv(RAW_PATH, low_memory=False)

rows_before = len(clean_df)

# Standardize text fields without changing category meaning
text_columns = clean_df.select_dtypes(include=["object", "str"]).columns

for col in text_columns:
    clean_df[col] = clean_df[col].str.strip()
    clean_df[col] = clean_df[col].replace("", pd.NA)

# Convert dates and times
clean_df["transaction_date"] = pd.to_datetime(
    clean_df["transaction_date"], errors="coerce"
)

clean_df["transaction_time"] = pd.to_datetime(
    clean_df["transaction_time"],
    format="%H:%M",
    errors="coerce"
).dt.time

# Add calendar fields for reporting
clean_df["transaction_year"] = clean_df["transaction_date"].dt.year
clean_df["transaction_month"] = clean_df["transaction_date"].dt.month
clean_df["transaction_year_month"] = (
    clean_df["transaction_date"].dt.to_period("M").astype("string")
)

# Add investigation flags; do not delete flagged records
clean_df["loan_type_missing_flag"] = (
    clean_df["has_loan"].eq(1) & clean_df["loan_type"].isna()
)

clean_df["high_value_flag"] = (
    clean_df["transaction_amount"] > 100_000
)

# Preserve all records; duplicates were already checked
assert len(clean_df) == rows_before
assert clean_df["transaction_id"].is_unique

# Save cleaned dataset
output_path = PROCESSED_DIR / "bank_transactions_cleaned.csv"
clean_df.to_csv(output_path, index=False)

# Save a basic quality report
quality_report = pd.DataFrame({
    "missing_count": clean_df.isna().sum(),
    "missing_pct": (
        clean_df.isna().mean().mul(100).round(2)
    )
}).sort_values("missing_pct", ascending=False)

quality_report.to_csv(
    REPORT_DIR / "data_quality_report.csv"
)

print("Cleaning completed.")
print("Rows preserved:", len(clean_df))
print("Columns after adding analysis fields:", len(clean_df.columns))
print("Output:", output_path.resolve())
print("Quality report:", (REPORT_DIR / "data_quality_report.csv").resolve())
print("\nRemaining missing values:")
display(quality_report.head(10))
```
